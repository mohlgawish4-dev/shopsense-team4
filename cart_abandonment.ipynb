```python
import pandas as pd 
```

```python
data = pd.read_csv("online_shoppers_intention.csv")
```

```python
data.head()
```

```python
print(data.shape)
print(data.columns)
```

```python
X = data.drop(columns=['Revenue'])
Y = data['Revenue']
print ("X shape: ", X.shape)
print ("Y shape: ", Y.shape )
```

```python
print(X.dtypes)
```

```python
X_encoded = pd.get_dummies(X , drop_first = True )
print ("X old shape: ",X.shape )
print("X new shape: ",X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(
    X_encoded, Y, test_size=0.2, random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train) 
print ("This model is trained successfully")
```

```python
predictions = model.predict(X_test)
print("first 10 predictions answers:", predictions[:10])
print("first 10 real answers:",Y_test.values[:10])
```

```python
from sklearn.metrics import accuracy_score 
acc=accuracy_score(Y_test,predictions)
print(f"model's accuracy : {acc*100:.2f}%")
```

```python
from sklearn.metrics import confusion_matrix 
cm =confusion_matrix(Y_test , predictions)
print("Confusion Matrix:")
print(cm)
```
